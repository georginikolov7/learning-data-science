#### 01. Data import:

```python
import pandas as pd
from pandas import DataFrame

#Import data from local file:
wine_data = pd.read_csv('../data/winemag-data-130k-v2.csv', index_col=0)
wine_data.head()
```

#### 02. Indexing:

```python
#4th row:
print(wine_data.iloc[5])

#third to 5th row (works same as in standard python):
print(wine_data.iloc[2:5])

#all to the 7th row:
print(wine_data.iloc[:7])

#LOC uses inclusive indices on both ends!!!:
#2nd, 3rd, 4th wine select the country and points:
print(wine_data.loc[2:4, ['country', 'points']])


#conditional selection:
print(wine_data.loc[wine_data.country == 'Spain'])

print(wine_data.loc[(wine_data.country == 'Italy') & (wine_data.points >= 90)])

#select values with null price (important to deal with null values when training ML models:
print(wine_data.loc[wine_data.price.isnull(),['country', 'points', 'price']])


#assign data: Eg. backwards index:
wine_data['index_backwards'] = range(len(wine_data), 0, -1)
wine_data['index_backwards']
```

#### 02.Maps:

```python
# eg impute all 0 and nan values to the mean using only pandas:
retail_df = pd.read_csv('../data/melb_data.csv')


def set_first_n_to_null(df,n:int,col_name:str):
    df.loc[:n-1, col_name]=None

#set price of the first 10 entries to null as dataset has no null values:
set_first_n_to_null(retail_df,10,'Price')

#use map to set them to mean:
mean = retail_df.Price.mean()
print(F'mean: {mean}')

retail_df.Price=retail_df.Price.map(lambda l: mean if pd.isna(l) else l)

print("Df with imputed prices:")
print(retail_df.loc[:9,'Price'])

#-------------------------------------------
#or of course use the fillna() in this case:
set_first_n_to_null(retail_df,10,'Price')

print('Df with null prices:')
print(retail_df.loc[:9,'Price'])

retail_df.Price=retail_df.Price.fillna(mean)
print("Df with imputed prices:")
print(retail_df.loc[:9,'Price'])
```
